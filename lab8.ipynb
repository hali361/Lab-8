### Normal case · Column selection

The input is the full DataFrame and the two wage column names. Expect 57 rows and two columns. The first two wage pairs should be `(5.0, NaN)` and `(4.5, 5.8)`. The multi-row `.loc[]` and `.iloc[]` selections in Task 2 should produce identical DataFrames.

```python
wage_columns = ["wage-increase-first-year", "wage-increase-second-year"]
expected_preview = pd.DataFrame(
    [[5.0, float("nan")], [4.5, 5.8]], columns=wage_columns
)
actual = df[["wage-increase-first-year", "wage-increase-second-year"]]

print(f"Input columns: {wage_columns}")
print(f"Expected shape: (57, 2) | Actual shape: {actual.shape}")
print("Expected first two rows")
display(expected_preview)
print("Actual first two rows")
display(actual.head(2))

by_label = df.loc[0:4, ["wage-increase-first-year", "wage-increase-second-year"]]
by_position = df.iloc[0:5, 1:3]
same_selection = by_label.equals(by_position)
print(f"Expected loc/iloc agreement: True | Actual: {same_selection}")

assert actual.shape == (57, 2)
pd.testing.assert_frame_equal(actual.head(2), expected_preview)
assert same_selection
print("PASS")
```

### Normal case · Outcome and numeric filters

Apply the five Task 3 filters to all 57 disputes. Compare the expected and actual number of rows for each condition.

```python
good = df[df["class"] == "good"]
bad = df[df["class"] == "bad"]
wage = df[df["wage-increase-first-year"] > 2.5]
holidays = df[df["statutory-holidays"] > 10]
both = df[(df["wage-increase-first-year"] > 2.5) &
          (df["statutory-holidays"] > 10)]

expected = [37, 20, 41, 37, 28]
actual = [len(good), len(bad), len(wage), len(holidays), len(both)]
display(pd.DataFrame({
    "Input condition": ["Good", "Bad", "Wage > 2.5", "Holidays > 10", "Both"],
    "Expected rows": expected,
    "Actual rows": actual,
}).set_index("Input condition"))

assert actual == expected
print("PASS")
```

### Special cases · One row, multiple rows, and no matches

Apply the combined Task 3 filter to three subsets. Rows 0 and 1 pass both conditions. Row 12 has exactly 10 holidays, and row 21 has exactly a 2.5% first-year wage increase. Each fails the corresponding comparison.

```python
filter_columns = ["wage-increase-first-year", "statutory-holidays", "class"]
print("Input rows used in the three subsets")
display(df.loc[[0, 1, 12, 21], filter_columns])

cases = [
    ("One row", [0, 12, 21], [0]),
    ("Multiple rows", [0, 1, 12, 21], [0, 1]),
    ("No rows", [12, 21], []),
]
results = []
for case, input_labels, expected_labels in cases:
    subset = df.loc[input_labels]
    matched = subset[(subset["wage-increase-first-year"] > 2.5) &
                     (subset["statutory-holidays"] > 10)]
    actual_labels = matched.index.tolist()
    results.append([case, input_labels, expected_labels, actual_labels, matched.shape])
    assert actual_labels == expected_labels
    assert matched.columns.equals(df.columns)

display(pd.DataFrame(results, columns=[
    "Case", "Input row labels", "Expected matches", "Actual matches", "Actual shape",
]).set_index("Case"))
print("PASS")
```

### Missing values · Filtering

`df.loc[[2, 3]]` selects the two rows by their index labels. The first table shows the input. Row 2 has a missing first-year wage increase and 11 statutory holidays. Row 3 has a 3.7% wage increase and a missing holiday count. Pandas displays both missing numeric values as `NaN`.

Each `>` comparison creates a Boolean mask with one value per row. Applying a mask to `subset` keeps only rows where the mask is `True`. In these numeric columns, comparing `NaN` with a threshold using `>` gives `False`.

| Row label | Wage increase > 2.5 | Statutory holidays > 10 | Both conditions |
| --- | --- | --- | --- |
| 2 | `False` because the wage is missing | `True` because 11 > 10 | `False` |
| 3 | `True` because 3.7 > 2.5 | `False` because holidays are missing | `False` |

The wage filter keeps row 3, and the holiday filter keeps row 2. The `&` operator combines the two masks row by row. A row survives only if both comparisons are `True` for that same row, so the combined filter returns an empty DataFrame.

The second table uses `.index.tolist()` to show the original row labels returned by each filter. The expected lists are `[3]`, `[2]`, and `[]`, in that order. The first two assertions confirm which values are missing, and the final assertion checks that all three actual lists match the expected lists.

```python
subset = df.loc[[2, 3]]
display(subset[filter_columns])

wage = subset[subset["wage-increase-first-year"] > 2.5]
holidays = subset[subset["statutory-holidays"] > 10]
both = subset[(subset["wage-increase-first-year"] > 2.5) &
              (subset["statutory-holidays"] > 10)]

expected = [[3], [2], []]
actual = [wage.index.tolist(), holidays.index.tolist(), both.index.tolist()]
display(pd.DataFrame({
    "Input condition": ["Wage > 2.5", "Holidays > 10", "Both"],
    "Expected row labels": expected,
    "Actual row labels": actual,
}).set_index("Input condition"))

assert pd.isna(subset.loc[2, "wage-increase-first-year"])
assert pd.isna(subset.loc[3, "statutory-holidays"])
assert actual == expected
print("PASS")
```

### Missing values · Mathematical operations

The input is the full first-year wage Series. The output also shows row label 2, where the wage is `NaN`.

- With the default `skipna=True`, `.mean()` and `.max()` skip the one missing wage. The expected mean is the sum of the known wages divided by their count (`213.0 / 56`), and the expected maximum is `7.0`.
- With `skipna=False`, both calculations return `NaN` because the Series contains a missing wage.
- `(wage_series + 1).loc[2]` tests arithmetic on the missing value itself. The result at row label 2 is still `NaN`.

```python
wage_series = df["wage-increase-first-year"]
print(f"Input: {wage_series.name}")
print("Input value at row label 2")
display(wage_series.loc[[2]])
print(f"Known wages: {wage_series.count()} | Missing wages: {wage_series.isna().sum()}")

expected = [213.0 / 56, 7.0, float("nan"), float("nan"), float("nan")]
actual = [
    df["wage-increase-first-year"].mean(),
    df["wage-increase-first-year"].max(),
    wage_series.mean(skipna=False),
    wage_series.max(skipna=False),
    (wage_series + 1).loc[2],
]
display(pd.DataFrame({
    "Operation": ["mean()", "max()", "mean(skipna=False)", "max(skipna=False)", "Wage at row 2 + 1"],
    "Expected": expected,
    "Actual": actual,
}).set_index("Operation"))

assert wage_series.count() == 56
pd.testing.assert_series_equal(pd.Series(actual), pd.Series(expected), check_exact=False)
print("PASS")
```
